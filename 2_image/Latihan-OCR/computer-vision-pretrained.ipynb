# TIMM Pretrained Models - Download Pipeline

Pipeline untuk download dan save pretrained models dari TIMM untuk penggunaan offline.

```python
import timm
import torch
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

SAVE_DIR = Path('pretrained_models')
SAVE_DIR.mkdir(exist_ok=True)

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Device: {device}")
print(f"TIMM version: {timm.__version__}")
print(f"Save directory: {SAVE_DIR.absolute()}")
```

```python
def download_and_save_model(model_name, save_dir):
    try:
        model = timm.create_model(model_name, pretrained=True)
        save_path = save_dir / f"{model_name}.pth"
        torch.save(model.state_dict(), save_path)
        print(f"Saved: {model_name}")
        return True
    except Exception as e:
        print(f"Failed: {model_name} - {str(e)}")
        return False

print("Download function ready")
```

## 13. Load Offline Model Example

Cara menggunakan model yang sudah di-download untuk offline usage

```python
def load_offline_model(model_name, num_classes=1000, model_dir='pretrained_models'):
    model = timm.create_model(model_name, pretrained=False, num_classes=num_classes)
    
    weight_path = Path(model_dir) / f"{model_name}.pth"
    if weight_path.exists():
        state_dict = torch.load(weight_path, map_location=device)
        model.load_state_dict(state_dict)
        print(f"Loaded weights from {weight_path}")
    else:
        print(f"Weight file not found: {weight_path}")
    
    return model

usage_example = """
USAGE EXAMPLE:

# Binary Classification (2 classes)
model = load_offline_model('efficientnet_b0', num_classes=2)

# Multi-class Classification (20 classes)
model = load_offline_model('resnet50', num_classes=20)

# Multi-label Classification (use sigmoid + BCE loss)
model = load_offline_model('vit_base_patch16_224', num_classes=10)

# For fine-tuning on custom dataset
model = model.to(device)
model.train()

# For inference
model.eval()
with torch.no_grad():
    output = model(input_tensor)
"""

print(usage_example)
```
