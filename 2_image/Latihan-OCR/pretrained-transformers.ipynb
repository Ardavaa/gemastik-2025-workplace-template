# Pretrained Vision Models Documentation - Hugging Face Transformers

Dokumentasi komprehensif untuk berbagai pretrained model vision dari Hugging Face Transformers.

Kategori:
1. Binary Classification
2. Multi-class Classification
3. Multi-label Classification
4. Object Detection
5. Semantic Segmentation
6. Image Segmentation
7. Zero-Shot Classification

```python
from transformers import (
    AutoModelForImageClassification, 
    AutoImageProcessor,
    AutoModelForObjectDetection,
    AutoModelForSemanticSegmentation,
    AutoModelForImageSegmentation,
    pipeline
)
from PIL import Image
import torch
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Device: {device}")
```

## 1. Binary Classification

Model untuk klasifikasi dua kelas (binary)

```python
binary_models = {
    "ViT": "google/vit-base-patch16-224",
    "DeiT": "facebook/deit-base-patch16-224",
    "BEiT": "microsoft/beit-base-patch16-224",
    "ConvNeXT": "facebook/convnext-base-224",
    "Swin": "microsoft/swin-base-patch4-window7-224",
    "ResNet": "microsoft/resnet-50",
    "EfficientNet": "google/efficientnet-b0"
}

def binary_classification_pipeline(model_name, image_path):
    processor = AutoImageProcessor.from_pretrained(model_name)
    model = AutoModelForImageClassification.from_pretrained(model_name).to(device)
    
    image = Image.open(image_path)
    inputs = processor(images=image, return_tensors="pt").to(device)
    
    with torch.no_grad():
        outputs = model(**inputs)
        logits = outputs.logits
        predicted_class = logits.argmax(-1).item()
        probabilities = torch.nn.functional.softmax(logits, dim=-1)
    
    return predicted_class, probabilities

print("Binary Classification Models:")
for name, model_id in binary_models.items():
    print(f"  - {name}: {model_id}")
```

## 2. Multi-class Classification

Model untuk klasifikasi banyak kelas (multi-class)

```python
multiclass_models = {
    "ViT-ImageNet": "google/vit-base-patch16-224",
    "ViT-ImageNet21k": "google/vit-base-patch16-224-in21k",
    "DeiT-ImageNet": "facebook/deit-base-distilled-patch16-224",
    "BEiT-ImageNet": "microsoft/beit-base-patch16-224-pt22k-ft22k",
    "Swin-ImageNet": "microsoft/swin-base-patch4-window7-224-in22k",
    "ConvNeXT-ImageNet": "facebook/convnext-base-224-22k",
    "EfficientNet-B7": "google/efficientnet-b7",
    "MobileViT": "apple/mobilevit-small",
    "RegNet": "facebook/regnet-y-040",
    "DenseNet": "facebook/deit-base-patch16-224"
}

def multiclass_classification_pipeline(model_name, image_path, top_k=5):
    classifier = pipeline("image-classification", model=model_name, device=0 if device=="cuda" else -1)
    results = classifier(image_path, top_k=top_k)
    return results

print("Multi-class Classification Models:")
for name, model_id in multiclass_models.items():
    print(f"  - {name}: {model_id}")
```

## 3. Zero-Shot Classification

Model untuk klasifikasi tanpa training (zero-shot)

```python
zeroshot_models = {
    "CLIP-ViT-Base": "openai/clip-vit-base-patch32",
    "CLIP-ViT-Large": "openai/clip-vit-large-patch14",
    "CLIP-ViT-Base-336": "openai/clip-vit-base-patch16",
    "Chinese-CLIP": "OFA-Sys/chinese-clip-vit-base-patch16",
    "ALIGN": "kakaobrain/align-base"
}

def zeroshot_classification_pipeline(model_name, image_path, candidate_labels):
    classifier = pipeline("zero-shot-image-classification", model=model_name, device=0 if device=="cuda" else -1)
    results = classifier(image_path, candidate_labels=candidate_labels)
    return results

print("Zero-Shot Classification Models:")
for name, model_id in zeroshot_models.items():
    print(f"  - {name}: {model_id}")
```

## 4. Object Detection

Model untuk deteksi objek dalam gambar

```python
object_detection_models = {
    "DETR-ResNet50": "facebook/detr-resnet-50",
    "DETR-ResNet101": "facebook/detr-resnet-101",
    "Conditional-DETR": "microsoft/conditional-detr-resnet-50",
    "YOLOS-Tiny": "hustvl/yolos-tiny",
    "YOLOS-Small": "hustvl/yolos-small",
    "YOLOS-Base": "hustvl/yolos-base",
    "DETA": "jozhang97/deta-swin-large",
    "Table-Transformer": "microsoft/table-transformer-detection",
    "DETR-DC5": "facebook/detr-resnet-50-dc5"
}

def object_detection_pipeline(model_name, image_path, threshold=0.5):
    detector = pipeline("object-detection", model=model_name, device=0 if device=="cuda" else -1)
    results = detector(image_path, threshold=threshold)
    return results

print("Object Detection Models:")
for name, model_id in object_detection_models.items():
    print(f"  - {name}: {model_id}")
```

## 5. Semantic Segmentation

Model untuk segmentasi semantik pixel-level

```python
semantic_segmentation_models = {
    "SegFormer-B0": "nvidia/segformer-b0-finetuned-ade-512-512",
    "SegFormer-B5": "nvidia/segformer-b5-finetuned-ade-640-640",
    "Mask2Former-Swin": "facebook/mask2former-swin-large-ade-semantic",
    "BEiT-Semantic": "microsoft/beit-base-finetuned-ade-640-640",
    "SegFormer-CityScapes": "nvidia/segformer-b0-finetuned-cityscapes-1024-1024",
    "UPerNet-ConvNeXT": "openmmlab/upernet-convnext-small"
}

def semantic_segmentation_pipeline(model_name, image_path):
    segmenter = pipeline("image-segmentation", model=model_name, device=0 if device=="cuda" else -1)
    results = segmenter(image_path)
    return results

print("Semantic Segmentation Models:")
for name, model_id in semantic_segmentation_models.items():
    print(f"  - {name}: {model_id}")
```

## 6. Instance Segmentation

Model untuk segmentasi instance (objek individual)

```python
instance_segmentation_models = {
    "Mask2Former-Swin-Base": "facebook/mask2former-swin-base-coco-instance",
    "Mask2Former-Swin-Large": "facebook/mask2former-swin-large-coco-instance",
    "MaskFormer-Swin": "facebook/maskformer-swin-base-coco",
    "OneFormer-ADE20K": "shi-labs/oneformer-ade20k-swin-large",
    "OneFormer-COCO": "shi-labs/oneformer-coco-swin-large"
}

def instance_segmentation_pipeline(model_name, image_path):
    processor = AutoImageProcessor.from_pretrained(model_name)
    model = AutoModelForImageSegmentation.from_pretrained(model_name).to(device)
    
    image = Image.open(image_path)
    inputs = processor(images=image, return_tensors="pt").to(device)
    
    with torch.no_grad():
        outputs = model(**inputs)
    
    return outputs

print("Instance Segmentation Models:")
for name, model_id in instance_segmentation_models.items():
    print(f"  - {name}: {model_id}")
```
